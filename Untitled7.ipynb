```python
# ==============================================================================
# ANÁLISIS EXPLORATORIO DE DATOS (EDA) Y VISUALIZACIÓN - GESTIÓN DE TURNOS
# Entorno: Google Colab / Jupyter Notebook
# ==============================================================================

# 1. INSTALACIÓN E IMPORTACIÓN DE LIBRERÍAS
try:
    import gdown
except ImportError:
    !pip install -q gdown
    import gdown

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración estética y de advertencias
warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_context("notebook", font_scale=1.1)
plt.rcParams['figure.dpi'] = 120
plt.rcParams['axes.edgecolor'] = '#CCCCCC'
plt.rcParams['axes.linewidth'] = 0.8

# ==============================================================================
# 2. DESCARGA Y CARGA ROBUSTA DE DATOS DESDE GOOGLE DRIVE
# ==============================================================================
FILE_ID = "1f0mq4PiB6peI8LCboTf-GRpusoFn3Vg4"
OUTPUT_FILE = "dataset_turnos.csv"

if not os.path.exists(OUTPUT_FILE):
    print("Descargando archivo desde Google Drive...")
    gdown.download(id=FILE_ID, output=OUTPUT_FILE, quiet=False)

def cargar_csv_robusto(ruta):
    """Intenta leer el CSV probando codificaciones y delimitadores comunes."""
    codificaciones = ['utf-8', 'latin1', 'cp1252']
    delimitadores = [';', '\t', ',']

    for enc in codificaciones:
        for sep in delimitadores:
            try:
                df_temp = pd.read_csv(ruta, sep=sep, encoding=enc, low_memory=False)
                # Validamos que se haya dividido en múltiples columnas correctamente
                if df_temp.shape[1] > 10:
                    print(f"✅ Archivo cargado con éxito | Encoding: '{enc}' | Delimitador: '{repr(sep)}'")
                    return df_temp
            except Exception:
                continue
    raise ValueError("No se pudo parsear el archivo automáticamente. Revisa los permisos del enlace.")

df = cargar_csv_robusto(OUTPUT_FILE)

# Limpieza de espacios en blanco en los nombres de las columnas
df.columns = df.columns.str.strip()

# ==============================================================================
# 3. LIMPIEZA, INGENIERÍA DE CARACTERÍSTICAS Y AUDITORÍA EDA
# ==============================================================================
def tiempo_a_minutos(serie):
    """Convierte strings 'HH:MM:SS' a minutos decimales de forma vectorizada."""
    td = pd.to_timedelta(serie.astype(str).str.strip(), errors='coerce')
    return td.dt.total_seconds() / 60.0

# Conversión de tiempos a minutos
df['Espera_min'] = tiempo_a_minutos(df['Espera'])
df['Atencion_min'] = tiempo_a_minutos(df['Atencion'])
df['Total_min'] = tiempo_a_minutos(df['Total'])

# Extracción de la hora del día (0 a 23) desde HoraSolicitud
horas_dt = pd.to_datetime(df['HoraSolicitud'].astype(str).str.strip(), format='%H:%M:%S', errors='coerce')
# Fallback por si algunas horas vienen sin segundos (HH:MM)
horas_dt = horas_dt.fillna(pd.to_datetime(df['HoraSolicitud'].astype(str).str.strip(), format='%H:%M', errors='coerce'))
df['Hora_Num'] = horas_dt.dt.hour

# Conversión numérica de día del mes
df['DiaSolicitud'] = pd.to_numeric(df['DiaSolicitud'], errors='coerce')

# Resumen Estadístico Ejecutivo en Consola
print("\n" + "="*70)
print("📊 RESUMEN EJECUTIVO DEL DATASET (EDA)")
print("="*70)
print(f"• Total de registros (Turnos): {len(df):,}")
print(f"• Total de columnas:           {df.shape[1]}")
print(f"• Periodo (Año - Mes):         {df['AñoSolicitud'].dropna().unique()} - Mes(es): {df['MesSolicitud'].dropna().unique()}")
print("\n--- Tiempos de Espera vs. Atención (en minutos) ---")
print(f"• Espera   -> Media: {df['Espera_min'].mean():.2f} min | Mediana: {df['Espera_min'].median():.2f} min | P90: {df['Espera_min'].quantile(0.90):.2f} min")
print(f"• Atención -> Media: {df['Atencion_min'].mean():.2f} min | Mediana: {df['Atencion_min'].median():.2f} min | P90: {df['Atencion_min'].quantile(0.90):.2f} min")
print("\n--- Distribución de Estado del Turno ---")
print(df['Estado'].value_counts(dropna=False))
print("="*70 + "\n")

# ==============================================================================
# 4. VISUALIZACIÓN 1: DEMANDA POR DÍA DEL MES Y FRANJA HORARIA
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 5.5))

# 4A. Demanda por Día del Mes
demanda_dia = df.groupby('DiaSolicitud')['IdTurno'].count().reset_index()
sns.lineplot(
    data=demanda_dia, x='DiaSolicitud', y='IdTurno',
    marker='o', linewidth=2.5, color='#1f77b4', ax=axes[0]
)
axes[0].fill_between(demanda_dia['DiaSolicitud'], demanda_dia['IdTurno'], color='#1f77b4', alpha=0.12)
axes[0].axhline(demanda_dia['IdTurno'].mean(), color='#d62728', linestyle='--', linewidth=1.2, label=f"Promedio diario ({demanda_dia['IdTurno'].mean():.0f})")

# Resaltar picos máximos
top_dias = demanda_dia.nlargest(2, 'IdTurno')
for _, row in top_dias.iterrows():
    axes[0].annotate(
        f"Día {int(row['DiaSolicitud'])}: {int(row['IdTurno'])}",
        xy=(row['DiaSolicitud'], row['IdTurno']),
        xytext=(0, 10), textcoords='offset points',
        ha='center', fontweight='bold', fontsize=9, color='#1f77b4'
    )

axes[0].set_title('Demanda Diaria de Turnos (Por Día del Mes)', fontweight='bold', pad=12)
axes[0].set_xlabel('Día del Mes')
axes[0].set_ylabel('Cantidad de Turnos')
axes[0].set_xticks(range(1, 32, 2))
axes[0].legend()

# 4B. Demanda por Franja Horaria
demanda_hora = df.groupby('Hora_Num')['IdTurno'].count().reset_index()
sns.lineplot(
    data=demanda_hora, x='Hora_Num', y='IdTurno',
    marker='s', linewidth=2.5, color='#2ca02c', ax=axes[1]
)
axes[1].fill_between(demanda_hora['Hora_Num'], demanda_hora['IdTurno'], color='#2ca02c', alpha=0.12)

for _, row in demanda_hora.iterrows():
    axes[1].text(row['Hora_Num'], row['IdTurno'] + (demanda_hora['IdTurno'].max()*0.02),
                 f"{int(row['IdTurno'])}", ha='center', fontsize=8.5, color='#333333')

axes[1].set_title('Afluencia por Franja Horaria (Dimensionamiento de Personal)', fontweight='bold', pad=12)
axes[1].set_xlabel('Hora del Día (24h)')
axes[1].set_ylabel('Cantidad de Turnos')
axes[1].set_xticks(demanda_hora['Hora_Num'].dropna().astype(int))
axes[1].set_xticklabels([f"{int(h):02d}:00" for h in demanda_hora['Hora_Num'].dropna()], rotation=45)

plt.tight_layout()
plt.show()

# ==============================================================================
# 5. VISUALIZACIÓN 2: TIEMPOS DE ESPERA VS. ATENCIÓN POR TRÁMITE
# ==============================================================================
# Seleccionamos los Top 10 trámites con mayor volumen para mantener legibilidad
top_tramites = df['Tramite_o_Servicio'].value_counts().nlargest(10).index
df_top_tramites = df[df['Tramite_o_Servicio'].isin(top_tramites)]

# Calculamos tanto Media (sensible a outliers) como Mediana (valor típico)
tiempos_tramite = df_top_tramites.groupby('Tramite_o_Servicio').agg(
    Espera_Media=('Espera_min', 'mean'),
    Atencion_Media=('Atencion_min', 'mean'),
    Espera_Mediana=('Espera_min', 'median'),
    Atencion_Mediana=('Atencion_min', 'median'),
    Volumen=('IdTurno', 'count')
).sort_values(by='Espera_Media', ascending=True)

# Acortar etiquetas muy largas para el eje Y
etiquetas_cortas = [t[:42] + '...' if len(str(t)) > 42 else str(t) for t in tiempos_tramite.index]

fig, axes = plt.subplots(1, 2, figsize=(17, 6.5), sharey=True)
y_pos = np.arange(len(tiempos_tramite))
altura_barra = 0.38

# 5A. Promedio (Detecta impacto de casos atípicos largos)
axes[0].barh(y_pos + altura_barra/2, tiempos_tramite['Espera_Media'], height=altura_barra, label='Espera (Promedio)', color='#ff7f0e')
axes[0].barh(y_pos - altura_barra/2, tiempos_tramite['Atencion_Media'], height=altura_barra, label='Atención (Promedio)', color='#1f77b4')
axes[0].set_yticks(y_pos)
axes[0].set_yticklabels(etiquetas_cortas, fontsize=9.5)
axes[0].set_title('Tiempos PROMEDIO por Trámite (Minutos)\n(Sensible a casos atípicos largos)', fontweight='bold')
axes[0].set_xlabel('Minutos')
axes[0].legend(loc='lower right')

# Etiquetas de valor en barras promedio
for i, (esp, ate) in enumerate(zip(tiempos_tramite['Espera_Media'], tiempos_tramite['Atencion_Media'])):
    axes[0].text(esp + 0.2, i + altura_barra/2, f"{esp:.1f}m", va='center', fontsize=8.5)
    axes[0].text(ate + 0.2, i - altura_barra/2, f"{ate:.1f}m", va='center', fontsize=8.5)

# 5B. Mediana (Comportamiento típico del 50% de los ciudadanos)
axes[1].barh(y_pos + altura_barra/2, tiempos_tramite['Espera_Mediana'], height=altura_barra, label='Espera (Mediana)', color='#ffbb78')
axes[1].barh(y_pos - altura_barra/2, tiempos_tramite['Atencion_Mediana'], height=altura_barra, label='Atención (Mediana)', color='#aec7e8')
axes[1].set_title('Tiempos MEDIANA por Trámite (Minutos)\n(Experiencia típica sin sesgo de extremos)', fontweight='bold')
axes[1].set_xlabel('Minutos')
axes[1].legend(loc='lower right')

for i, (esp, ate) in enumerate(zip(tiempos_tramite['Espera_Mediana'], tiempos_tramite['Atencion_Mediana'])):
    axes[1].text(esp + 0.1, i + altura_barra/2, f"{esp:.1f}m", va='center', fontsize=8.5)
    axes[1].text(ate + 0.1, i - altura_barra/2, f"{ate:.1f}m", va='center', fontsize=8.5)

plt.tight_layout()
plt.show()

# ==============================================================================
# 6. VISUALIZACIÓN 3: EFECTIVIDAD Y DESERCIÓN POR TIPO DE CLIENTE
# ==============================================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 6A. Barras 100% Apiladas: Estado del Turno segmentado por Tipo de Ciudadano
ct_estado = pd.crosstab(df['Ciudadano'].fillna('Sin Especificar'), df['Estado'].fillna('Sin Estado'))
ct_estado_pct = ct_estado.div(ct_estado.sum(axis=1), axis=0) * 100

colores_estado = {'Finalizado': '#2ca02c', 'Abandonado': '#d62728'}
paleta_6a = [colores_estado.get(col, '#7f7f7f') for col in ct_estado_pct.columns]

ct_estado_pct.plot(
    kind='bar', stacked=True, ax=axes[0], color=paleta_6a, width=0.6, edgecolor='white'
)
axes[0].set_title('Tasa de Efectividad vs. Abandono por Tipo de Ciudadano (%)', fontweight='bold', pad=12)
axes[0].set_xlabel('Segmento de Ciudadano')
axes[0].set_ylabel('Porcentaje de Turnos (%)')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=0)
axes[0].legend(title='Estado', bbox_to_anchor=(1.01, 1), loc='upper left')

# Añadir etiquetas con porcentaje y conteo absoluto dentro de las barras
for idx, ciudadano in enumerate(ct_estado_pct.index):
    acumulado = 0
    for estado in ct_estado_pct.columns:
        pct = ct_estado_pct.loc[ciudadano, estado]
        conteo = ct_estado.loc[ciudadano, estado]
        if pct > 3.5:  # Solo mostrar texto si la barra tiene suficiente altura
            axes[0].text(
                idx, acumulado + pct / 2,
                f"{pct:.1f}%\n({conteo:,})",
                ha='center', va='center', color='white', fontweight='bold', fontsize=9
            )
        acumulado += pct

# 6B. Gráfico de Anillo (Donut): Resultado del Trámite ((SERVICIO) Resultado del trámite)
col_resultado = '(SERVICIO) Resultado del trámite'
if col_resultado in df.columns:
    res_counts = df[col_resultado].fillna('Sin Registro / Abandonado').value_counts()
    # Agrupar categorías minoritarias (< 1.5%) en "Otros" para limpiar el anillo
    umbral = len(df) * 0.015
    res_principales = res_counts[res_counts >= umbral]
    if (res_counts < umbral).any():
        res_principales['Otros'] = res_counts[res_counts < umbral].sum()

    colores_donut = sns.color_palette('Set2', len(res_principales))
    wedges, texts, autotexts = axes[1].pie(
        res_principales,
        labels=res_principales.index,
        autopct='%1.1f%%',
        startangle=140,
        colors=colores_donut,
        pctdistance=0.78,
        wedgeprops=dict(width=0.42, edgecolor='white', linewidth=2)
    )
    for autotext in autotexts:
        autotext.set_color('#222222')
        autotext.set_fontweight('bold')
        autotext.set_fontsize(9.5)

    # Texto central del anillo
    axes[1].text(0, 0, f"Total\n{len(df):,}\nTurnos", ha='center', va='center', fontweight='bold', fontsize=11)
    axes[1].set_title('Distribución del Resultado del Trámite', fontweight='bold', pad=12)

plt.tight_layout()
plt.show()
```
